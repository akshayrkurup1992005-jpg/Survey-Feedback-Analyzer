## Step 1: Initialize Survey Feedback Data

```python
import pandas as pd
import re

feedback_data = {
    "S_No": list(range(1, 11)),
    "Name": ["Ravi", "Meera", "Sam", "Anu", "Raj",
             "Divya", "Arjun", "Kiran", "Leela", "Nisha"],
    "Feedback": [
        "   Very GOOD Service!!!",
        "poor support,        not happy   ",
        "GREAT experience! will come again.",
        "okay     okay...",
        "   not    BAD",
        "Excellent care, excellent staff!",
        "good food and good ambience!",
        "Poor response and poor handling of issue",
        "Satisfied. But could be better.",
        "Good support... quick service."
    ],
    "Rating": [5, 2, 5, 3, 2, 5, 4, 1, 3, 4]
}

display(pd.DataFrame(feedback_data))
```

## Step 2: Collect Additional Feedbacks

```python
n = int(input("Enter number of additional feedbacks: "))

for _ in range(n):
    name = input("Name: ")
    comment = input("Feedback: ")

    while True:
        try:
            rating = int(input("Rating (1-5): "))
            if rating in range(1, 6):
                break
            print("Enter a rating between 1 and 5.")
        except ValueError:
            print("Enter a valid integer.")

    feedback_data["S_No"].append(len(feedback_data["S_No"]) + 1)
    feedback_data["Name"].append(name)
    feedback_data["Feedback"].append(comment)
    feedback_data["Rating"].append(rating)

display(pd.DataFrame(feedback_data))
```

## Step 3: Clean and Standardize Feedback

```python
for i, comment in enumerate(feedback_data["Feedback"]):
    comment = re.sub(r"[.,!?]", "", comment)
    comment = re.sub(r"\s+", " ", comment).strip()
    feedback_data["Feedback"][i] = comment.lower()

display(pd.DataFrame(feedback_data))
```

## Step 4: Analyze Frequently Mentioned Words

```python
def count_word_in_feedbacks(word):
    return sum(
        word.lower() in comment.split()
        for comment in feedback_data["Feedback"]
    )

words_to_check = ["good", "poor", "excellent"]

word_summary = pd.DataFrame({
    "Word": words_to_check,
    "Feedback Count": [
        count_word_in_feedbacks(word)
        for word in words_to_check
    ]
})

display(word_summary)
```

## Step 5: Final Summary and Insights

```python
df = pd.DataFrame(feedback_data)

# Display final cleaned data
display(df)

# Calculate average rating
average = df["Rating"].mean()

rating_summary = pd.DataFrame({
    "Total Feedbacks": [len(df)],
    "Average Rating": [round(average, 2)]
})

display(rating_summary)

# Find the longest feedback
df["Word_Count"] = df["Feedback"].apply(
    lambda text: len(text.split())
)

longest = df.loc[[df["Word_Count"].idxmax()]]

display(longest[["S_No", "Name", "Feedback", "Word_Count"]])

# Extract unique words
unique_words = sorted({
    word
    for comment in df["Feedback"]
    for word in comment.split()
})

display(pd.DataFrame({"Unique Words": unique_words}))
```
